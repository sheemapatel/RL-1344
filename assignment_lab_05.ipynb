```python
import gymnasium as gym
import random


def run_random_agent(env_name, episodes=10):

    print("\n" + "=" * 50)
    print("Environment:", env_name)
    print("=" * 50)

    # Create environment
    env = gym.make(env_name)

    episode_rewards = []

    for episode in range(episodes):

        # Reset environment
        state, info = env.reset()

        total_reward = 0
        terminated = False
        truncated = False

        while not (terminated or truncated):

            # Select a random action
            action = env.action_space.sample()

            # Execute action
            next_state, reward, terminated, truncated, info = env.step(action)

            # Add reward
            total_reward += reward

            # Move to next state
            state = next_state

        episode_rewards.append(total_reward)

        print(
            "Episode", episode + 1,
            "Total Reward:", total_reward
        )

    # Calculate average reward
    average_reward = sum(episode_rewards) / len(episode_rewards)

    print("-" * 50)
    print("Total Reward:", sum(episode_rewards))
    print("Average Reward:", average_reward)

    env.close()


# ------------------------------------------------
# Run Random Agent on all three environments
# ------------------------------------------------

run_random_agent("CartPole-v1", episodes=10)

run_random_agent("MountainCar-v0", episodes=10)

run_random_agent("FrozenLake-v1", episodes=10)
```
